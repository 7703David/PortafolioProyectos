# Número Par o Impar

```python
x = float(input("Ingrese el número: "))
if x % 2 == 0:
    print("El número {x} es par")
else:
    print("El número {x} es impar")
```

```python
es_par_o_impar = lambda x: f"El número {x} es par" if x % 2 == 0 else f"El número {x} es impar"

x = float(input("Ingrese el número: "))
print(es_par_o_impar(x))
```

#Ejercicio 5
suma de los números desde 1 hasta n.

```python
n = int(input("Introduce un número entero positivo: "))

if n <= 0:
    print("El número debe ser mayor que cero.")
else:
    suma = 0
    for i in range(1, n + 1):
        suma += i
    print(f"La suma de los números de 1 a {n} es: {suma}")
```

#Ejercicio 6
Imprimir el número más grande de una lista

```python
entrada = input("Introduce una lista de números separados por comas: ")

numeros = [float(num.strip()) for num in entrada.split(",")]

if not numeros:
    print("La lista está vacía.")
else:
    maximo = numeros[0]

    for num in numeros:
        if num > maximo:
            maximo = num

    print(f"El número máximo en la lista es: {maximo}")
```

#Ejercicio 7
Fibonacci

```python
n = int(input("¿Cuántos términos de la serie de Fibonacci deseas ver? "))

if n <= 0:
    print("Por favor, introduce un número positivo.")
else:
    fibonacci = []
    a, b = 0, 1
    for i in range(n):
        fibonacci.append(a)
        a, b = b, a + b
    print(f"Serie de Fibonacci: {fibonacci}")
```
